## About Dataset

This dataset contains the full set of country-level rankings and happiness scores from the World Happiness Update
Report 2026, released on March 20, 2026. It includes data for 147 countries spanning all major regions of the world.

Each entry corresponds to a single country and reports its overall happiness score along with six key explanatory
factors: income (economic prosperity), social support, life expectancy (health), freedom to make life choices, generosity, and perceived corruption.

The data is drawn from the Gallup World Poll, which conducts nationally representative surveys in more than 140 
countries annually. To ensure greater stability and comparability, the scores in this edition are averaged over the 
2023–2025 period, minimizing short-term fluctuations.

This dataset is well-suited for a range of analytical tasks, including exploratory data analysis (EDA), correlation 
analysis, predictive modeling, and geospatial visualization. It can also be integrated with other datasets—such as GDP, 
population statistics, or digital behavior metrics—for more comprehensive research.

A key highlight of the 2026 report is its focus on social media and youth wellbeing. The findings suggest that young 
adults in English-speaking countries have seen notable declines in life satisfaction over the past two decades, with 
researchers pointing to high levels of social media use as a contributing factor.

This project uses the World Happiness Report 2026 dataset to explore and understand differences in happiness among countries. Various charts such as bar charts, line charts, pie charts, scatter plots, and a correlation heatmap are used to turn numerical data into clear visual stories.

# 🎯 Interview Questions — Data Cleaning & Preprocessing

## 1. How do you decide whether to drop or impute a missing value?

### Answer:

I decide based on the **amount of missing data, importance of the column, and reason for the missing value**.

- If only a few rows have missing values and they are not important, I may **drop those rows**.
- If the column is important and has a small number of missing values, I may **impute** them using mean, median, or mode.
- For numerical data with outliers, **median** is often safer than mean.
- If missingness itself is meaningful, I may **flag it** instead of replacing it.
- If a column has too many missing values and provides little analytical value, I may consider **dropping the column**.

### Example:

For my World Happiness dataset, all columns had **0 missing values**, so no dropping or imputation was required.

## 2. What's the difference between a duplicate row and a duplicate key?

### Answer:

A **duplicate row** means the entire record is repeated with the same values across all columns.

A **duplicate key** means the value that should uniquely identify a record appears more than once, even if the other column values are different.

## 3. How would you handle outliers differently from missing values?

### Answer:

Missing values and outliers are different data-quality problems, so I would not treat them the same way.

- **Missing value:** A value is absent. I may **drop, impute, or flag** it depending on the situation.
- **Outlier:** A value is present but unusually high or low compared with the rest of the data.

For outliers, I would first investigate whether they are:

- A genuine observation
- A data-entry error
- A measurement problem

I can use methods such as **IQR, Z-score, box plots, or domain-specific limits**.

I would **not automatically remove an outlier**, because it may represent an important real-world observation.

---

## 4. How would you validate that a dataset is "clean" before starting analysis?

### Answer:

I would perform a final data-quality checklist before analysis:

1. Check the **number of rows and columns**.
2. Check **missing values**.
3. Check **duplicate rows**.
4. Check **duplicate keys/identifiers**.
5. Verify **data types**.
6. Standardize **text formatting and whitespace**.
7. Check **numerical ranges and invalid values**.
8. Investigate **outliers**.
9. Verify **date formats and units**, if applicable.
10. Re-run all inventory checks after cleaning.
11. Review a sample of the cleaned data.
12. Export and save the final cleaned dataset.
