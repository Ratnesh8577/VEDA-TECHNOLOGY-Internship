# 📊 Exploratory Data Analysis — Sample Superstore

## 📊 About the Dataset and Analysis

### About the Dataset

The **Sample Superstore Dataset** is a retail sales dataset that contains detailed information about customer orders, products, sales transactions, shipping methods, geographical locations, discounts, and profits. It is commonly used for learning and demonstrating **Exploratory Data Analysis (EDA), data visualization, and business analytics**.

The dataset originally contains **9,994 records and 13 columns**. The columns provide information about shipping, customers, locations, products, sales, quantity, discounts, and profit. Important categorical columns include **Ship Mode, Segment, Country, City, State, Region, Category, and Sub-Category**. The numerical columns include **Postal Code, Sales, Quantity, Discount, and Profit**.

This dataset is useful for understanding how different business factors influence sales and profitability. It allows analysts to investigate questions such as which categories generate the highest revenue, which regions perform best, how discounts affect profit, and which products or sub-categories are generating losses.

### About the Analysis

The main purpose of this project is to perform a complete **Exploratory Data Analysis (EDA)** before drawing business conclusions. The analysis begins with loading and inspecting the dataset using **Python, Pandas, NumPy, Matplotlib, and Seaborn**.

The initial analysis checks the dataset shape, column names, data types, missing values, and duplicate records. The dataset contains **no missing values**, while **17 duplicate records** were identified. After removing these duplicates, the final dataset contains **9,977 unique records**.

Next, descriptive statistics are calculated using `describe()` to understand the distribution of numerical variables such as Sales, Quantity, Discount, and Profit. The analysis shows that sales values are highly right-skewed because a small number of transactions have very high sales values.

Several visualizations are created to understand the data more clearly. **Histograms** are used to study Sales, Profit, and Quantity distributions. **Boxplots** are used to identify potential outliers. A **correlation heatmap** is created to understand relationships between numerical variables. **Bar charts** are used to compare sales and profit across categories, sub-categories, regions, and customer segments. **Scatter plots** are used to investigate relationships such as Sales vs Profit and Discount vs Profit.

The analysis shows that the business generated approximately **$2.30 million in sales and $286,000 in profit**, with an overall profit margin of approximately **12.5%**. Technology is the strongest category in terms of both sales and profit, while Furniture has a much lower profit margin. The **Tables** sub-category is particularly concerning because it generates a significant overall loss.

Another important finding is the relationship between discount and profitability. Higher discounts are generally associated with lower profits, and very high discount levels can result in negative profitability. The analysis also shows a moderate positive relationship between Sales and Profit, but high sales do not always guarantee high profit.

Overall, this EDA project demonstrates how statistical analysis and visualization can transform raw retail data into meaningful **business insights and recommendations**. It provides a strong foundation for further analysis, reporting, and data-driven decision-making.

## Statistical Outlier Detection — IQR

```python
Q1 = df_clean["Sales"].quantile(0.25)
Q3 = df_clean["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
sales_outliers = df_clean[
    (df_clean["Sales"] < lower_bound) |
    (df_clean["Sales"] > upper_bound)
]

sales_outliers.shape
```

# 🎯 Interview Questions — Exploratory Data Analysis

## 1. How do you distinguish a genuine outlier from a data entry error?

### Answer:

I distinguish a genuine outlier from a data entry error by checking the **business context, source data, and related columns**.

* First, I identify the outlier using methods such as **IQR or Z-score**.
* I check whether the value is logically possible.
* I compare it with related columns such as **Quantity, Sales, Discount, and Profit**.
* I check the original/source data to confirm whether the value was entered correctly.
* If the value represents a real business transaction, I **keep it**.
* If it is caused by a typo, incorrect unit, or invalid entry, I **correct or remove it** based on the situation.

**Key Point:**
I never remove an outlier automatically just because it is statistically unusual.

---

## 2. What's the difference between correlation and causation, and why does it matter in EDA?

### Answer:

**Correlation** means that two variables are statistically related, while **causation** means that a change in one variable directly causes a change in another variable.

### Example:

In this dataset:

* **Discount and Profit** have a negative correlation.
* This means higher discounts are generally associated with lower Profit.
* However, correlation alone does **not prove** that Discount is the only reason for the decrease in Profit.
* Other factors such as **product cost, pricing, category, and region** may also affect Profit.

### Why does it matter?

It matters because confusing **correlation with causation** can lead to incorrect conclusions and poor business decisions.

**Key Point:**

> Correlation shows a relationship, but it does not necessarily mean one variable causes the other.

---

## 3. Which chart type would you use to show a trend over 12 months, and why?

I would use a **Line Chart** to show a trend over 12 months.

### Reasons:

* A line chart is designed to show **changes over time**.
* It clearly shows **increases, decreases, peaks, and fluctuations**.
* Months can be arranged chronologically on the X-axis.
* It makes it easier to identify **seasonal patterns and trends**.
* It allows us to compare performance across different months.

### Example:

I could use a line chart to display **monthly Sales or monthly Profit** and identify which months performed better or worse.

**Key Point:**

> A line chart is the best choice for showing trends and changes over a continuous time period.
