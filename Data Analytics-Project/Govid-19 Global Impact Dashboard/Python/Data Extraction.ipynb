```python
#Importing Required Libraries:
import pandas as pd
import requests
import json
```

# Time Series (State) Data:

```python
#Making a Get Request to the Server:
response=requests.get("https://data.covid19india.org/v4/min/timeseries.min.json")
#Converting the response to Python DataStructure and Storing it to a Variable:
data=response.json()
```

```python
data
```

```python
#Making Empty DataFrame:
df = pd.DataFrame()
#Iterating through Each State:
for state_code, state_data in data.items():
    #Getting All the Dates and Storing in a Variable:
    timeseries_data = state_data.get('dates', {})
    #Iterating through Each District of the State:
    for date, date_data in timeseries_data.items():
        #Flattening the Dictionary:
        date_df = pd.json_normalize(date_data)
        #Storing Dates in One Varaibel:
        date_df['Date'] = date
        #Storing Satate_Code in One Varaibel:
        date_df['State_Code'] = state_code
        #Concatinating Each Dates's Data
        df = pd.concat([df, date_df], ignore_index=True)
```

```python
#Displaying All the Columns:
pd.set_option("display.max_columns",None)
```

```python
df
```

```python
#Checking for Noise, and Unwanted Columns:
for i in df:
    print(f"Column: {i}\n---------------\nValues:\n{df[i].value_counts()}\n----------------\n\n")
```

```python
#Storing All The Unwanted Columns in a Variable:
to_drop=["total.other","delta7.other","delta.other"]
```

```python
#Droping Unwanted Columns:
df.drop(to_drop,axis=1,inplace=True)
```

```python
#Renaming and Rearranging Columns:
df=df[column_order_and_names.keys()].rename(columns=column_order_and_names)
```

```python
#Filling the Null Values with 0:
for i in df:
    df[i]=df[i].fillna(0)
    #Changing The DataTypes:
    try:
        if i=="Date":
            df[i]=pd.to_datetime(df[i],format="%Y-%m-%d")
        else:    
            df[i]=df[i].astype(int)
    except:
        df[i]
```

```python
#Removing Outliers:
for i in df.select_dtypes(include=int):
    q1=df[i].quantile(0.25)
    q3=df[i].quantile(0.75)
    iqr=q3-q1
    ul=df[i]+iqr*1.5
    ll=df[i]-iqr*1.5
    df=df[(df[i]>=ll)&(df[i]<=ul)]
```

```python
#Making CSV File to Use it In Different Tools:
df.to_csv("TimeSeries_Data.csv", index=False)
```

# District Data:

```python
#Making a Get Request to the Server:
response=requests.get("https://data.covid19india.org/v4/min/data.min.json")
#Converting the response to Python DataStructure and Storing it to a Variable:
data=response.json()
data
```

```python
#Making Empty DataFrame:
df = pd.DataFrame()
#Iterating through Each State:
for state_code, state_data in data.items():
    #Getting All the Districts of the State in a Variable:
    districts_data = state_data.get('districts', {})
    #Iterating through Each District of the State:
    for district_code, district_data in districts_data.items():
        #Flattening the Dictionary:
        district_df = pd.json_normalize(district_data)
        #Storing Satate_Code in One Varaibel:
        district_df['State_Code'] = state_code
        #Storing District in One Varaibel:
        district_df['District'] = district_code
        #Concatinating all the Individual District Data into one DataFrame:
        df = pd.concat([df,district_df], ignore_index=True)
```

```python
#Displaying All the Columns:
pd.set_option("display.max_columns",None)
```
