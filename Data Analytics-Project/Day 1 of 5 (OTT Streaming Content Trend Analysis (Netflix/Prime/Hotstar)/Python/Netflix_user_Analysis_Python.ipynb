Q2. What is the total revenue generated?

Q3. What is the Average Revenue Per User (ARPU)?

```python
# Q3. What is the Average Revenue Per User (ARPU)?

total_users = df["User_ID"].nunique()
total_revenue = df["Monthly_Subscription_Fee"].sum()

arpu = total_revenue / total_users

print(f"Average Revenue Per User (ARPU): ${arpu:,.2f}")
```

 Business Insight

- The Average Revenue Per User (ARPU) is $6,587.27.
- ARPU measures how much revenue each unique user contributes on average.
- This KPI helps evaluate customer value and can be used to compare the performance of different subscription plans and user segments.

Q4. Which Subscription Plan Generates the Highest Revenue?

```python
# Q4. Which Subscription Plan Generates the Highest Revenue?

import matplotlib.pyplot as plt

# Group by Subscription Plan
plan_revenue = (
    df.groupby("Subscription_Plan")["Monthly_Subscription_Fee"]
    .sum()
    .sort_values(ascending=True)
)

# Create Figure
plt.figure(figsize=(8,5))

# Horizontal Bar Chart
plt.barh(
    plan_revenue.index,
    plan_revenue.values
)

# Title & Labels
plt.title("Revenue by Subscription Plan", fontsize=14, fontweight="bold")
plt.xlabel("Total Revenue ($)", fontsize=12)
plt.ylabel("Subscription Plan", fontsize=12)

# Show values on bars
for index, value in enumerate(plan_revenue.values):
    plt.text(value, index, f'${value:,.0f}', va='center')

plt.tight_layout()
plt.show()
```

```python
highest_plan = plan_revenue.idxmax()
highest_revenue = plan_revenue.max()

print(f"Highest Revenue Plan: {highest_plan}")
print(f"Revenue: ${highest_revenue:,.2f}")
```

Business Insight

- The Premium subscription plan generated the highest total revenue, contributing $15,422,400.40.
- This indicates that Premium subscribers play a significant role in the platform's overall revenue.
- Netflix can increase revenue further by encouraging Standard and Basic users to upgrade to the Premium plan through personalized offers and additional features.

Q5. Which Countries Contribute the Most Revenue?


```python
# Q5. Which Countries Contribute the Most Revenue?

import matplotlib.pyplot as plt

# Group revenue by country
country_revenue = (
    df.groupby("Country_of_User")["Monthly_Subscription_Fee"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

# Plot
plt.figure(figsize=(10,6))

plt.barh(
    country_revenue.index[::-1],
    country_revenue.values[::-1]
)

plt.title("Top 10 Countries by Revenue", fontsize=14, fontweight="bold")
plt.xlabel("Total Revenue ($)", fontsize=12)
plt.ylabel("Country", fontsize=12)

# Add value labels
for index, value in enumerate(country_revenue.values[::-1]):
    plt.text(value, index, f'${value:,.0f}', va='center', fontsize=10)

plt.tight_layout()
plt.show()
```

 Business Insight

- The top 10 countries contribute the largest share of the platform's subscription revenue.
- High-revenue countries represent the most valuable customer markets and should be prioritized for customer retention and premium subscription campaigns.
- Countries with lower revenue may require localized marketing strategies or pricing adjustments to increase subscriber growth.

Q7. Which Genres Have the Highest Total Watch Time?

Q8. How Does Total Watch Time Vary by Month?

```python
# Q8. Monthly Watch Time Trend

month_watch = (
    df.groupby("Month")["Watch_Time_Minutes"]
      .sum()
      .reindex([
          "January","February","March","April","May","June",
          "July","August","September","October","November","December"
      ])
)

plt.figure(figsize=(12,6))

plt.plot(month_watch.index, month_watch.values, marker='o')

plt.title("Monthly Watch Time Trend", fontsize=14, fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Watch Time (Minutes)")

plt.xticks(rotation=45)

plt.grid(True)

plt.tight_layout()
plt.show()
```

Business Insight

- Monthly watch time reveals seasonal viewing patterns.
- Peak viewing months may coincide with holidays, vacations, or major content releases.
- Understanding these trends helps optimize content release schedules and marketing campaigns.

Q14. Which Genres Receive the Highest Average User Ratings?

```python
# Q14. Which Genres Receive the Highest Average User Ratings?

genre_rating = (
    df.groupby("Genre")["Rating_Given"]
      .mean()
      .sort_values(ascending=False)
)

print(genre_rating)

plt.figure(figsize=(10,6))

plt.bar(
    genre_rating.index,
    genre_rating.values
)

plt.title("Average User Rating by Genre", fontsize=14, fontweight="bold")
plt.xlabel("Genre")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)

for index, value in enumerate(genre_rating.values):
    plt.text(index, value, f'{value:.2f}', ha='center')

plt.tight_layout()
plt.show()
```

Business Insight

- Highly rated genres indicate stronger viewer satisfaction and content quality.
- These genres should be prioritized for future content acquisition and production.
- Lower-rated genres may benefit from improved recommendations or higher-quality content.
